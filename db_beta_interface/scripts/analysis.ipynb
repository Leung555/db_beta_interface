```python
import math
import numpy as np
```

```python
def find_closest_angle(a1, a2):
    if a1 > a2:
        a1, a2 = a2, a1
    ans_1 = abs(a2-a1)
    ans_2 = abs((a1+360)-a2)
    return min(ans_1, ans_2)
```

```python
num_leg = 6
leg_angle = []
for i in range(num_leg):
    rand_angle = np.random.randint(-180, 180)
    leg_angle.append(rand_angle)
leg_angle = sorted(leg_angle)
print('leg_angle: ', leg_angle)

leg_angle = [0, 60, 120, -180, -120, -60]
leg_cover_range = []
for i in range(len(leg_angle)):
    # print(i)
    if i == len(leg_angle)-1:
        Range = find_closest_angle(leg_angle[i-1], leg_angle[0])
    else:
        Range = find_closest_angle(leg_angle[i-1], leg_angle[i+1])
    # print("index left: ", leg_angle[i-1], "   index right: ", leg_angle[i+1])
    leg_cover_range.append(Range)
print('leg_cover_range: ', leg_cover_range)
print('Sum angle: ', sum(leg_cover_range))
```
